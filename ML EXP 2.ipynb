```python
import pandas as pd

data = pd.DataFrame([
    ['Big', 'Red', 'Circle', 'No'],
    ['Small', 'Red', 'Triangle', 'No'],
    ['Small', 'Red', 'Circle', 'Yes'],
    ['Big', 'Blue', 'Circle', 'No'],
    ['Small', 'Blue', 'Circle', 'Yes']
], columns=['Size', 'Color', 'Shape', 'Class'])

print("Training Data:")
print(data)

attributes = list(data.columns[:-1])
S = ['0'] * len(attributes)
G = [['?' for _ in attributes]]

for _, row in data.iterrows():
    x = list(row[:-1])
    label = row['Class']

    if label == 'Yes':
        G = [g for g in G if all(
            g[i] == '?' or g[i] == x[i]
            for i in range(len(x))
        )]

        for i in range(len(S)):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        new_G = []
        for g in G:
            if all(g[i] == '?' or g[i] == x[i]
                   for i in range(len(x))):
                for i in range(len(g)):
                    if g[i] == '?':
                        for value in data[attributes[i]].unique():
                            if value != x[i] and S[i] == value:
                                h = g.copy()
                                h[i] = value
                                new_G.append(h)
            else:
                new_G.append(g)
        G = new_G

print("\nSpecific Boundary (S):", S)
print("General Boundary (G):")
for h in G:
    print(h)
```
