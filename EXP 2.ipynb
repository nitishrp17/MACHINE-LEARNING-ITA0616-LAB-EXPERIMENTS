```python
import pandas as pd

data = pd.read_csv("training_data.csv")
print("Training Data:\n", data)

attributes = list(data.columns[:-1])
S = ['0'] * len(attributes)
G = [['?' for _ in attributes]]

for _, row in data.iterrows():
    x = list(row[:-1])
    label = str(row.iloc[-1]).strip()

    if label == 'Yes':
        G = [g for g in G if all(
            g[i] == '?' or g[i] == x[i]
            for i in range(len(x))
        )]

        for i in range(len(S)):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        new_G = []

        for g in G:
            if all(g[i] == '?' or g[i] == x[i]
                   for i in range(len(x))):
                for i in range(len(g)):
                    if g[i] == '?':
                        for value in data[attributes[i]].unique():
                            if value != x[i] and (
                                S[i] == '?' or S[i] == value
                                or S[i] == '0'
                            ):
                                h = g.copy()
                                h[i] = value

                                if all(
                                    S[j] == '0' or S[j] == '?'
                                    or h[j] == '?'
                                    or h[j] == S[j]
                                    for j in range(len(S))
                                ):
                                    new_G.append(h)
            else:
                new_G.append(g)

        G = new_G

print("\nSpecific Boundary (S):", S)
print("General Boundary (G):")
for h in G:
    print(h)
```
